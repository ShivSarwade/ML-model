# Phase 1.5: Data CleaningHere we handle nulls, fix data types, format strings, and recalculate the target percentage.

```python
import pandas as pd
import numpy as np
```

## 1. Load Data

```python
df = pd.read_csv('../data/raw/attendance_raw.csv')
```

## 2. Cleaning Function

```python
def clean_data(df):
    df_clean = df.copy()
    # Drop nulls in crucial targets
    df_clean = df_clean.dropna(subset=['Students_Present', 'Total_Enrolled'])
    
    # Cap just in case
    df_clean['Students_Present'] = np.minimum(df_clean['Students_Present'], df_clean['Total_Enrolled'])
    
    # Datetime formatting
    df_clean['Date'] = pd.to_datetime(df_clean['Date'], format='%d-%m-%Y', errors='coerce')
    
    # Recalculate target
    df_clean['Attendance_Percentage'] = (df_clean['Students_Present'] / df_clean['Total_Enrolled']) * 100
    
    # Sort chronologically for temporal features later
    df_clean = df_clean.sort_values(by=['Date', 'Lecture_Number']).reset_index(drop=True)
    return df_clean

df_cleaned = clean_data(df)
df_cleaned.to_csv('../data/processed/attendance_cleaned.csv', index=False)
```
