# Phase 1: Data Collection and ValidationIn this notebook, we load the raw attendance dataset and perform critical logical validations.

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import logging

logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')
```

## 2. Load Raw Dataset

```python
raw_path = '../data/raw/attendance_raw.csv'
df_raw = pd.read_csv(raw_path)
df_raw.head()
```

## 3. Validation LogicWe check for missing columns, impossible bounds (Students > Enrolled), and null values.

```python
def validate_data(df):
    required_cols = ['Date', 'Subject', 'Total_Enrolled', 'Students_Present', 'Start_Time', 'Internal_Test_Week']
    missing = [c for c in required_cols if c not in df.columns]
    if missing:
        logging.error(f'Missing: {missing}')
        return False
    
    invalid = df[df['Students_Present'] > df['Total_Enrolled']]
    if not invalid.empty:
        logging.warning(f'Found {len(invalid)} records where Students Present > Enrolled.')
        
    logging.info('Validation complete.')
    return True

is_valid = validate_data(df_raw)
```
