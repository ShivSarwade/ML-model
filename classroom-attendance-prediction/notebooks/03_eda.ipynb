# Phase 2: Exploratory Data Analysis (EDA)Visualizing the distribution of attendance and identifying trends.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

os.makedirs('../visualizations', exist_ok=True)
```

```python
df = pd.read_csv('../data/processed/attendance_cleaned.csv')
df['Date'] = pd.to_datetime(df['Date'])
```

## 1. Overall Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Attendance_Percentage'], bins=20, kde=True)
plt.title('Distribution of Attendance Percentage')
plt.xlabel('Attendance %')
plt.ylabel('Frequency')
plt.savefig('../visualizations/attendance_distribution.png')
plt.show()
```
